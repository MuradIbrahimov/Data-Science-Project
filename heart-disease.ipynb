```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# -------------------------------
# 2. Load dataset
# -------------------------------
path = r"C:\Users\murad\.cache\kagglehub\datasets\oktayrdeki\heart-disease\versions\1\heart_disease.csv"
df = pd.read_csv(path)

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

# Preview first rows
display(df.head())

# -------------------------------
# 3. Basic information
# -------------------------------
print("\n--- Dataset Info ---")
df.info()

print("\n--- Missing Values ---")
print(df.isna().sum())

print("\n--- Descriptive Statistics ---")
display(df.describe())

# -------------------------------
# 4. Check for duplicates
# -------------------------------
dup_count = df.duplicated().sum()
print(f"\nNumber of duplicate rows: {dup_count}")

# -------------------------------
# 5. Basic visual exploration
# -------------------------------

# Histogram of all numeric features
df.hist(figsize=(15, 10), bins=20)
plt.suptitle("Feature Distributions", fontsize=16)
plt.show()

# -------------------------------
# Correlation Heatmap (numeric columns only)
# -------------------------------

numeric_df = df.select_dtypes(include=['float64', 'int64'])

plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), annot=False, cmap="coolwarm")
plt.title("Correlation Heatmap (Numeric Features Only)")
plt.show()


# Target balance (assuming 'target' column exists)
if "target" in df.columns:
    plt.figure(figsize=(6,4))
    sns.countplot(data=df, x="target")
    plt.title("Target Variable Distribution")
    plt.show()
else:
    print("\nNo 'target' column found.")
```
